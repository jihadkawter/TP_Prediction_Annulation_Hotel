## .1 Collecte des données

**Principe important :** on ne modifie jamais le fichier source. On le lit, on le contrôle,
puis on en sauvegarde une **copie brute intouchée** dans `Donnees/raw/`. Tout le travail de nettoyage
se fera ensuite sur des copies dérivées, jamais sur cette version brute — ce qui garantit la
**traçabilité** des données à tout moment du projet.

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
sns.set_style('whitegrid')

PROJECT_ROOT = Path.cwd().resolve()
if not (PROJECT_ROOT / "data").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
SOURCE_FILE = PROJECT_ROOT / "data" / "source" / "hotel_bookings.csv"
RAW_FILE = PROJECT_ROOT / "data" / "raw" / "hotel_bookings_raw.csv"

print("Racine du projet :", PROJECT_ROOT)
print("Fichier source    :", SOURCE_FILE)
```

```python
df = pd.read_csv(SOURCE_FILE)
print(f"Dimensions de la source : {df.shape[0]:,} lignes x {df.shape[1]} colonnes")
df.head()
```

### Vérification de la structure des données

Cette étape permet de vérifier les types de variables ainsi que le nombre de valeurs non nulles avant d'entreprendre l'analyse exploratoire.

```python
df.info()
```

### 1.1 Contrôles techniques minimaux

```python
# Contrôles de base avant toute analyse
assert "is_canceled" in df.columns, "La variable cible is_canceled est absente."
assert df.shape[0] > 0, "Le fichier source est vide."
print("Variable cible détectée :", "is_canceled")

print(f"Nombre d'observations : {df.shape[0]}")
print(f"Nombre de variables : {df.shape[1]}")
```

### 1.2 Sauvegarde d'une copie brute (traçabilité)

```python
RAW_FILE.parent.mkdir(parents=True, exist_ok=True)
df.to_csv(RAW_FILE, index=False)
print("Copie brute enregistrée :", RAW_FILE)
print("Taille :", f"{RAW_FILE.stat().st_size / 1024**2:.2f} Mo")
```

# Conclusion

Le jeu de données a été chargé avec succès et une copie brute a été créée dans le dossier `data/raw`.

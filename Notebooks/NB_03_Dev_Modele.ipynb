# NB_03 · Développement des modèles
### Prédiction d'annulation de réservations d'hôtel

**Objectif :** entraîner et sauvegarder 3 modèles de classification supervisée sur les données
préparées dans `NB_02_Pretraitement` :
1. Logistic Regression (baseline)
2. Decision Tree
3. Random Forest


```python
import pandas as pd
import numpy as np
import joblib
import os

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

X_train = pd.read_csv('../Donnees/processed/X_train.csv')
X_test = pd.read_csv('../Donnees/processed/X_test.csv')
y_train = pd.read_csv('../Donnees/processed/y_train.csv').squeeze()
y_test = pd.read_csv('../Donnees/processed/y_test.csv').squeeze()

print("X_train :", X_train.shape)
print("X_test  :", X_test.shape)
```

## Modèle 1 · Logistic Regression

**Décision :** commencer par un modèle simple et rapide afin d'avoir une *baseline* pour comparer
les modèles plus complexes.

```python
log_reg = LogisticRegression(max_iter=1000, random_state=42)
log_reg.fit(X_train, y_train)

os.makedirs('../Modeles', exist_ok=True)
joblib.dump(log_reg, '../Modeles/logistic_regression.pkl')
print("Logistic Regression entraîné et sauvegardé.")
```

## Modèle 2 · Decision Tree

**Décision :** tester un arbre de décision pour apprendre des règles plus complexes / non linéaires
que la régression logistique.

```python
tree_clf = DecisionTreeClassifier(max_depth=10, random_state=42)
tree_clf.fit(X_train, y_train)

joblib.dump(tree_clf, '../Modeles/decision_tree.pkl')
print("Decision Tree entraîné et sauvegardé.")
```

## Modèle 3 · Random Forest

**Décision :** utiliser une méthode d'ensemble (agrégation de plusieurs arbres) pour réduire la
sensibilité à un seul arbre et améliorer la généralisation.

```python
rf_clf = RandomForestClassifier(n_estimators=300, random_state=42, n_jobs=-1)
rf_clf.fit(X_train, y_train)

joblib.dump(rf_clf, '../Modeles/random_forest.pkl')
print("Random Forest entraîné et sauvegardé.")
```

## Résumé

Les 3 modèles entraînés sont sauvegardés dans `../Modeles/` :
- `logistic_regression.pkl`
- `decision_tree.pkl`
- `random_forest.pkl`
